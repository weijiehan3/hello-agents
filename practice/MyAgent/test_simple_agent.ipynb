```python
from dotenv import load_dotenv
from hello_agents import HelloAgentsLLM, ToolRegistry
from hello_agents.tools import CalculatorTool
from agents.simple_agent import MySimpleAgent
from core.llm import MyLLM2
import logging

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# 加载环境变量
load_dotenv()


```

```python
# 创建LLM实例
llm = MyLLM2(provider="modelscope")

# 测试1: 基础对话Agent（无工具）
logger.info("=== 测试1: 基础对话Agent（无工具） ===")
basic_agent = MySimpleAgent(name="基础对话Agent", llm=llm, system_prompt="你是一个基础对话Agent，专注于与用户进行友好和有趣的对话。")

response1 = basic_agent.run("您好，请介绍一下你自己，你是什么模型，是由那家公司开发的？")
print("基础对话Agent的回答:", response1)
```

```python
# 测试2: 带工具的Agent
logger.info("=== 测试2: 带工具的Agent ===")
tool_registry = ToolRegistry()
calculator = CalculatorTool()
tool_registry.register_tool(calculator)

enhanced_agent = MySimpleAgent(
    name="带工具的Agent",
    llm=llm,
    system_prompt="你是一个智能助手，可以使用工具来帮助用户",
    tool_registry=tool_registry,
    enable_tool_calling=True
)

response2 =enhanced_agent.run("请帮我计算 12345 * 6789")
print("带工具的Agent的回答:", response2)
```
